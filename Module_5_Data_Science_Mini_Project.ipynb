## 2. Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
pd.set_option('display.max_columns', None)
```

## 3. Load the Real-World Dataset

The Titanic dataset is a publicly available real-world dataset containing passenger information and survival outcomes.

```python
url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)

print('Dataset loaded successfully!')
print('Rows:', df.shape[0])
print('Columns:', df.shape[1])
df.head()
```

## 4. Understand the Dataset

```python
print('Column names:')
print(df.columns.tolist())

print('\nDataset information:')
df.info()

print('\nSummary statistics:')
display(df.describe(include='all').T)
```

## 5. Check Missing Values

```python
missing = df.isnull().sum().sort_values(ascending=False)
print(missing[missing > 0])
```

## 6. Data Cleaning

```python
# Fill missing age values with the median age
df['Age'] = df['Age'].fillna(df['Age'].median())

# Fill missing embarked values with the most common value
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Remove the Cabin column because it has many missing values
df = df.drop(columns=['Cabin'])

print('Missing values after cleaning:')
print(df.isnull().sum()[df.isnull().sum() > 0])
```

## 7. Survival Overview

```python
survival_counts = df['Survived'].value_counts().sort_index()
print('0 = Did not survive, 1 = Survived')
print(survival_counts)

survival_rate = df['Survived'].mean() * 100
print(f'Overall survival rate: {survival_rate:.2f}%')
```

```python
plt.figure(figsize=(7,5))
sns.countplot(data=df, x='Survived')
plt.title('Titanic Survival Count')
plt.xlabel('Survived (0 = No, 1 = Yes)')
plt.ylabel('Number of Passengers')
plt.show()
```

## 8. Survival by Gender

```python
gender_survival = df.groupby('Sex')['Survived'].mean().sort_values(ascending=False) * 100
print(gender_survival.round(2))
```

```python
plt.figure(figsize=(7,5))
sns.barplot(data=df, x='Sex', y='Survived')
plt.title('Survival Rate by Gender')
plt.ylabel('Survival Rate')
plt.xlabel('Gender')
plt.ylim(0, 1)
plt.show()
```

## 9. Survival by Passenger Class

```python
class_survival = df.groupby('Pclass')['Survived'].mean() * 100
print(class_survival.round(2))
```

```python
plt.figure(figsize=(7,5))
sns.barplot(data=df, x='Pclass', y='Survived')
plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.ylim(0, 1)
plt.show()
```

## 10. Age Distribution

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x='Age', bins=30, kde=True)
plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.show()
```

## 11. Fare Distribution

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x='Fare', bins=30, kde=True)
plt.title('Fare Distribution')
plt.xlabel('Fare')
plt.ylabel('Number of Passengers')
plt.show()
```

## 12. Survival by Gender and Passenger Class

```python
plt.figure(figsize=(8,5))
sns.barplot(data=df, x='Pclass', y='Survived', hue='Sex')
plt.title('Survival Rate by Passenger Class and Gender')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.ylim(0, 1)
plt.show()
```

## 13. Correlation Heatmap

```python
numeric_cols = df.select_dtypes(include=np.number)
plt.figure(figsize=(9,6))
sns.heatmap(numeric_cols.corr(), annot=True, fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

## 14. Key Insights

- The dataset contains passenger information such as age, gender, passenger class, fare, and survival status.
- The overall survival rate can be calculated directly from the `Survived` column.
- Survival rates differ between male and female passengers.
- Passenger class is associated with different survival rates.
- Age and fare distributions show variation among passengers.
- The combined gender/class visualization helps identify differences in survival patterns.
- Correlation analysis provides a quick view of relationships among numerical variables.

**Note:** These are descriptive observations from this dataset and should not be interpreted as proof of causation.
